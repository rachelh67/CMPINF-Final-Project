Of the three metrics, my portion is to analyze the amount of farmers markets and farm stands in neighborhoods in Allegheny County. The "best" neighborhood for foodies being the one with the most farmer markets and stands.

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

#load data
df = pd.read_csv("farmers_markets.csv")
df.head(25)
```

**Clean data** because cities are typed out in caps or lowercase

```python
#Analyzing data
zip_counts = df['zip_code'].value_counts()
print(f"Number of unique zip codes: {len(zip_counts)} *out of 75")
```

**Assigning each zip code to a neighborhood**

```python
#Categorizing Data
neighborhood_map = {
    15037: "Elizabeth/Forward",
    15132: "McKeesport",
    15201: "Lawrenceville",
    15224: "Garfield/Bloomfield",
    15212: "Brighton Heights",
    15203: "South Side",
    15207: "Greenfield",
    15227: "Brentwood",
    15237: "Ross/The Block",
    15239: "Plum Borough",
    15106: "Carnegie",
    15221: "Wilkinsburg",
    15101: "Hampton/Allison Park",
    15090: "Wexford",
    15139: "Oakmont",
    15044: "Gibsonia",
    15208: "Homewood",
    15228: "Mt Lebanon",
    15136: "McKees Rocks",
    15122: "West Mifflin",
    15205: "Crafton",
    15007: "Bakerstown",
    15025: "Clairton/Jefferson Hills",
    15026: "Clinton",
    15031: "Cuddy",
    15057: "McDonald",
    15063: "Monongahela",
    15065: "Natrona Heights",
    15068: "New Kensington",
    15071: "Oakdale",
    15084: "Tarentum",
    15104: "Braddock/Rankin",
    15120: "Homestead/Munhall",
    15127: "Ingomar",
    15147: "Verona",
    16101: "New Castle",
}

df['neighborhood'] = df['zip_code'].map(neighborhood_map)
```

**Visualizing data**

```python
neighborhood_count = df['neighborhood'].value_counts()

plt.figure(figsize=(15, 10))
df['neighborhood'].value_counts().sort_values(ascending=True).plot(kind='barh', color='blue')
plt.title('Farmers Markets/Stands by Neighborhood')
plt.xlabel('Number of Locations')
plt.ylabel('Neighborhood')
plt.tight_layout()
plt.show()
```

```python
best_neighborhood = neighborhood_count.idxmax()
totalMarkets = neighborhood_count.max()

print(f"The best neighborhood is {best_neighborhood} with {totalMarkets} markets.")
```

**Conclusion** Based on the analysis of farmer markets and stands across Allegheny County, the "best" neighborhood for access to fresh produce and farmer stands is Wexford.
